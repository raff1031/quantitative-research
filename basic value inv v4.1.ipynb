```python
# comprehensive_comparison_enhanced.py - VERSIONE MIGLIORATA con CSV export e analisi trades
import os
import time
import math
import warnings
from dataclasses import dataclass
from datetime import datetime, timedelta
from typing import Dict, List, Optional, Tuple
from concurrent.futures import ThreadPoolExecutor, as_completed
import multiprocessing as mp

import numpy as np
import pandas as pd
import yfinance as yf
from tqdm import tqdm
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

warnings.filterwarnings("ignore", category=FutureWarning)

# ============================= CONFIG =================================

VERBOSE = True
YF_SLEEP = 0.02
MAX_WORKERS = min(8, mp.cpu_count())

@dataclass
class BacktestConfig:
    initial_capital: float = 10_000.0
    per_trade_usd: float = 200.0
    start_year: int = 2021
    rolling_window: int = 30
    min_periods: int = 15
    entry_k_std: float = -0.5
    exit_k_std: float = 0.5
    accumulate_daily: bool = True
    fractional_shares: bool = True
    max_alloc_per_ticker_frac: float = 0.05
    
    # Parametri per confronto
    rebalance_frequency_years: int = 2
    min_trading_days: int = 100
    min_market_cap_billion: float = 1.0
    
    # Larry Connors RSI(2) parameters
    rsi_period: int = 2
    rsi_entry_threshold: float = 20
    rsi_exit_threshold: float = 70
    rsi_ma_period: int = 200
    rsi_ma_type: str = "SMA"
    rsi_max_holding_days: int = 10
    rsi_profit_target_pct: float = 5.0
    
    # COMMISSIONI - ora più chiare e corrette
    commission_per_trade_usd: float = 0.0       # Commissione fissa per trade (es. 1.0 USD)
    commission_per_trade_pct: float = 0.0       # Commissione percentuale (es. 0.001 = 0.1%)
    
    random_seed: int = 42
    risk_free_rate: float = 0.02
    
    # CSV export settings
    export_trades_csv: bool = True
    csv_output_dir: str = "backtest_results"

# ============================= UNIVERSE ===================

REALISTIC_LARGE_CAP = [
    # Tech
    "AAPL","MSFT","AMZN","GOOGL","GOOG","ORCL","CSCO","INTC","IBM","QCOM","AMAT","ADI","NVDA","AMD",
    
    # Finance
    "JPM","BAC","WFC","GS","MS","C","USB","PNC","COF","AXP","BLK","SCHW",
    
    # Healthcare
    "JNJ","PFE","MRK","ABBV","LLY","UNH","CVS","BMY","ABT","GILD","AMGN","MDT","TMO","DHR","BSX","ISRG",
    
    # Consumer & Retail
    "WMT","HD","TGT","COST","LOW","KO","PEP","PG","MCD","SBUX","NKE","TJX","DIS",
    
    # Energy
    "XOM","CVX","COP","SLB","HAL","OXY","DVN","EOG",
    
    # Industrials
    "GE","CAT","MMM","HON","BA","LMT","NOC","GD","UPS","FDX","DE","EMR","ITW",
    
    # Materials
    "DD","DOW","LYB","APD","PPG","SHW","ECL","FCX","NEM",
    
    # Utilities
    "D","SO","NEE","DUK","EXC","ED","AEP","XEL",
    
    # REITs
    "AMT","PLD","PSA","EQR","AVB","UDR","CPT","MAA","ESS","BXP","SPG"
]

def get_realistic_universe() -> List[str]:
    return REALISTIC_LARGE_CAP

# ============================= UTILS ==================================

def today_str() -> str:
    return datetime.utcnow().strftime("%Y-%m-%d")

def calculate_rsi(prices: pd.Series, period: int = 14) -> pd.Series:
    """Calculate Relative Strength Index (RSI)"""
    delta = prices.diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=period).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=period).mean()
    
    rs = gain / loss
    rsi = 100 - (100 / (1 + rs))
    
    return rsi

def calculate_sma(series: pd.Series, period: int) -> pd.Series:
    """Calculate Simple Moving Average"""
    return series.rolling(window=period).mean()

def calculate_ema(series: pd.Series, period: int) -> pd.Series:
    """Calculate Exponential Moving Average"""
    return series.ewm(span=period, adjust=False).mean()

def calculate_drawdown(equity_series: pd.Series) -> pd.Series:
    """Calcola il drawdown della serie equity"""
    running_max = equity_series.expanding().max()
    drawdown = (equity_series / running_max - 1) * 100
    return drawdown

def calculate_max_drawdown(equity_series: pd.Series) -> float:
    """Calcola il massimo drawdown"""
    drawdown = calculate_drawdown(equity_series)
    return drawdown.min()

def calculate_sharpe_ratio(equity_series: pd.Series, risk_free_rate: float = 0.02) -> float:
    """Calcola lo Sharpe Ratio STANDARD: (Ra - Rf) / σ"""
    if len(equity_series) < 2:
        return 0.0
    
    returns = equity_series.pct_change().dropna()
    
    if len(returns) == 0 or returns.std() == 0:
        return 0.0
    
    annual_return = (1 + returns.mean()) ** 252 - 1
    annual_volatility = returns.std() * np.sqrt(252)
    
    sharpe = (annual_return - risk_free_rate) / annual_volatility
    
    return sharpe

def calculate_commission(trade_value: float, cfg: BacktestConfig) -> float:
    """
    Calcola la commissione totale per un trade
    
    Args:
        trade_value: Valore del trade in USD (positivo)
        cfg: Configurazione con parametri commissioni
    
    Returns:
        Commissione totale in USD
    """
    commission = cfg.commission_per_trade_usd + (trade_value * cfg.commission_per_trade_pct)
    return commission

# ========================= Data fetching (OTTIMIZZATO PER ANNO) ===============================

def fetch_prices_yf_single_yearly(args: Tuple[str, str, str]) -> Tuple[str, pd.Series]:
    """
    Download prezzi per singolo ticker, UN ANNO ALLA VOLTA per evitare rate limits
    """
    symbol, start_full, end_full = args
    
    try:
        start_date = pd.to_datetime(start_full)
        end_date = pd.to_datetime(end_full)
        
        # Divide in chunks annuali
        all_data = []
        current_start = start_date
        
        while current_start < end_date:
            # Calcola end per questo chunk (1 anno o fino alla fine)
            current_end = min(current_start + timedelta(days=365), end_date)
            
            # Download chunk
            try:
                chunk = yf.download(
                    symbol, 
                    start=current_start.strftime("%Y-%m-%d"), 
                    end=current_end.strftime("%Y-%m-%d"), 
                    progress=False, 
                    auto_adjust=True, 
                    threads=False
                )
                
                if chunk is not None and len(chunk) > 0:
                    all_data.append(chunk)
                
                # Sleep per evitare rate limit
                time.sleep(YF_SLEEP)
                
            except Exception as e:
                if VERBOSE:
                    print(f"  Errore chunk {symbol} {current_start.year}: {e}")
            
            # Prossimo anno
            current_start = current_end
        
        # Combina tutti i chunks
        if not all_data:
            return symbol, pd.Series(dtype=float)
        
        data = pd.concat(all_data)
        
        # Estrai Close
        close = data["Close"] if "Close" in data.columns else data["Adj Close"]
        if isinstance(close, pd.DataFrame):
            close = close.iloc[:, 0]
        
        close = pd.to_numeric(close, errors="coerce").dropna()
        close.index = pd.to_datetime(close.index).tz_localize(None)
        close.name = "price"
        
        return symbol, close
        
    except Exception as e:
        if VERBOSE:
            print(f"  Errore generale {symbol}: {e}")
        return symbol, pd.Series(dtype=float)

def fetch_prices_parallel(symbols: List[str], start: str, end: str) -> Dict[str, pd.Series]:
    """Download parallelo dei prezzi con chunking annuale"""
    print(f"Download parallelo di {len(symbols)} simboli con chunking annuale...")
    print(f"Periodo: {start} -> {end}")
    print(f"Workers: {MAX_WORKERS}")
    
    args_list = [(sym, start, end) for sym in symbols]
    results = {}
    
    with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
        future_to_symbol = {executor.submit(fetch_prices_yf_single_yearly, args): args[0] for args in args_list}
        
        for future in tqdm(as_completed(future_to_symbol), total=len(symbols), desc="Download prezzi"):
            symbol = future_to_symbol[future]
            try:
                _, price_series = future.result()
                if not price_series.empty and len(price_series) > 200:
                    results[symbol] = price_series
            except Exception as e:
                if VERBOSE:
                    print(f"Errore downloading {symbol}: {e}")
    
    print(f"✓ Scaricati con successo {len(results)}/{len(symbols)} simboli")
    return results

def fetch_yf_fundamentals_single(symbol: str) -> Tuple[str, pd.Series, float]:
    """Versione per parallelizzazione - scarica fondamentali singolo ticker"""
    try:
        ticker = yf.Ticker(symbol)
        
        info = ticker.info
        if not info:
            return symbol, pd.Series(dtype=float), 0.0
        
        market_cap = info.get("marketCap", 0.0)
        
        try:
            hist = ticker.history(period="5y", auto_adjust=True)
            if hist.empty or "Close" not in hist.columns:
                return symbol, pd.Series(dtype=float), market_cap
            
            close_prices = hist["Close"]
            close_prices.index = pd.to_datetime(close_prices.index).tz_localize(None)
            
            time.sleep(YF_SLEEP)
            return symbol, close_prices, market_cap
            
        except:
            return symbol, pd.Series(dtype=float), market_cap
            
    except Exception:
        return symbol, pd.Series(dtype=float), 0.0

def fetch_yf_fundamentals_parallel(symbols: List[str]) -> Tuple[Dict[str, pd.Series], Dict[str, float]]:
    """Download parallelo dei fondamentali"""
    print(f"Download fondamentali per {len(symbols)} simboli...")
    
    price_data = {}
    market_caps = {}
    
    with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
        future_to_symbol = {executor.submit(fetch_yf_fundamentals_single, sym): sym for sym in symbols}
        
        for future in tqdm(as_completed(future_to_symbol), total=len(symbols), desc="Download fondamentali"):
            try:
                symbol, prices, mcap = future.result()
                if not prices.empty:
                    price_data[symbol] = prices
                if mcap > 0:
                    market_caps[symbol] = mcap
            except Exception as e:
                if VERBOSE:
                    print(f"Errore fundamentals: {e}")
    
    return price_data, market_caps

# ========================= TRADE ANALYSIS ===============================

def analyze_trades_from_csv(csv_path: str, cfg: BacktestConfig) -> Dict:
    """
    Analizza un CSV di trades e calcola metriche chiave
    
    Returns:
        Dict con: max_win, max_loss, avg_trade, win_rate, profit_factor, sharpe_ratio
    """
    if not os.path.exists(csv_path):
        print(f"File CSV non trovato: {csv_path}")
        return {}
    
    df = pd.read_csv(csv_path)
    
    if df.empty or 'pnl_usd' not in df.columns:
        print(f"CSV vuoto o formato non valido")
        return {}
    
    # Filtra solo trades chiusi (con pnl)
    closed_trades = df[df['pnl_usd'].notna()].copy()
    
    if len(closed_trades) == 0:
        print("Nessun trade chiuso nel CSV")
        return {}
    
    pnls = closed_trades['pnl_usd'].values
    
    # Calcoli base
    max_win = pnls.max()
    max_loss = pnls.min()
    avg_trade = pnls.mean()
    
    # Win rate
    winning_trades = pnls[pnls > 0]
    losing_trades = pnls[pnls < 0]
    win_rate = len(winning_trades) / len(pnls) if len(pnls) > 0 else 0.0
    
    # Profit factor
    gross_profit = winning_trades.sum() if len(winning_trades) > 0 else 0.0
    gross_loss = abs(losing_trades.sum()) if len(losing_trades) > 0 else 0.0
    profit_factor = gross_profit / gross_loss if gross_loss > 0 else 0.0
    
    # Sharpe ratio dei trades (non dell'equity!)
    sharpe_trades = 0.0
    if len(pnls) > 1 and pnls.std() > 0:
        sharpe_trades = (pnls.mean() * np.sqrt(252)) / pnls.std()  # Annualizzato
    
    # Total PnL
    total_pnl = pnls.sum()
    
    # Average win / Average loss
    avg_win = winning_trades.mean() if len(winning_trades) > 0 else 0.0
    avg_loss = losing_trades.mean() if len(losing_trades) > 0 else 0.0
    
    results = {
        'total_trades': len(closed_trades),
        'winning_trades': len(winning_trades),
        'losing_trades': len(losing_trades),
        'max_win': max_win,
        'max_loss': max_loss,
        'avg_trade': avg_trade,
        'avg_win': avg_win,
        'avg_loss': avg_loss,
        'win_rate': win_rate,
        'profit_factor': profit_factor,
        'sharpe_ratio_trades': sharpe_trades,
        'total_pnl': total_pnl,
        'gross_profit': gross_profit,
        'gross_loss': gross_loss
    }
    
    return results

def print_trade_analysis(analysis: Dict, strategy_name: str):
    """Stampa analisi trades in formato leggibile"""
    if not analysis:
        print(f"\n=== ANALISI TRADES: {strategy_name} ===")
        print("Nessun dato disponibile")
        return
    
    print(f"\n=== ANALISI TRADES: {strategy_name} ===")
    print(f"Total Trades:          {analysis['total_trades']}")
    print(f"Winning Trades:        {analysis['winning_trades']} ({analysis['win_rate']*100:.1f}%)")
    print(f"Losing Trades:         {analysis['losing_trades']}")
    print(f"\nPROFIT/LOSS:")
    print(f"Max Win:               ${analysis['max_win']:,.2f}")
    print(f"Max Loss:              ${analysis['max_loss']:,.2f}")
    print(f"Average Trade:         ${analysis['avg_trade']:,.2f}")
    print(f"Average Win:           ${analysis['avg_win']:,.2f}")
    print(f"Average Loss:          ${analysis['avg_loss']:,.2f}")
    print(f"\nMETRICS:")
    print(f"Total P&L:             ${analysis['total_pnl']:,.2f}")
    print(f"Gross Profit:          ${analysis['gross_profit']:,.2f}")
    print(f"Gross Loss:            ${analysis['gross_loss']:,.2f}")
    print(f"Profit Factor:         {analysis['profit_factor']:.2f}")
    print(f"Sharpe (Trades):       {analysis['sharpe_ratio_trades']:.2f}")

# ========================= SIGNAL GENERATION ===============================

def generate_ps_ratio_signal(ticker: str, price_data: Dict[str, pd.Series], 
                             market_caps: Dict[str, float], cfg: BacktestConfig) -> Optional[pd.DataFrame]:
    """Generate P/S ratio signals for a single ticker"""
    if ticker not in price_data:
        return None
    
    prices = price_data[ticker].copy()
    if len(prices) < cfg.rolling_window + 20:
        return None
    
    # Get market cap
    market_cap = market_caps.get(ticker, 0.0)
    # Fix: verifica che non sia una Series
    if isinstance(market_cap, pd.Series) or market_cap <= 0:
        return None
    
    # Calcola rolling metrics
    roll_mean = prices.rolling(window=cfg.rolling_window, min_periods=cfg.min_periods).mean()
    roll_std = prices.rolling(window=cfg.rolling_window, min_periods=cfg.min_periods).std()
    
    z_score = (prices - roll_mean) / roll_std
    z_score = z_score.replace([np.inf, -np.inf], np.nan).fillna(0)
    
    # Genera segnali
    df = pd.DataFrame({
        'price': prices,
        'z_score': z_score,
        'signal': 0
    })
    
    df.loc[z_score < cfg.entry_k_std, 'signal'] = 1   # BUY
    df.loc[z_score > cfg.exit_k_std, 'signal'] = -1   # SELL
    
    return df

def generate_rsi2_signal(ticker: str, price_data: Dict[str, pd.Series], 
                        market_caps: Dict[str, float], cfg: BacktestConfig) -> Optional[pd.DataFrame]:
    """
    Generate RSI(2) Larry Connors signals for a single ticker
    """
    if ticker not in price_data:
        return None
    
    prices = price_data[ticker].copy()
    if len(prices) < cfg.rsi_ma_period + cfg.rsi_period + 20:
        return None
    
    # Calcola RSI(2)
    rsi = calculate_rsi(prices, period=cfg.rsi_period)
    
    # Calcola Moving Average per trend filter
    if cfg.rsi_ma_type == "EMA":
        ma = calculate_ema(prices, period=cfg.rsi_ma_period)
    else:
        ma = calculate_sma(prices, period=cfg.rsi_ma_period)
    
    # Genera segnali
    df = pd.DataFrame({
        'price': prices,
        'rsi': rsi,
        'ma': ma,
        'signal': 0
    })
    
    # BUY: RSI < threshold AND price > MA (uptrend)
    buy_condition = (rsi < cfg.rsi_entry_threshold) & (prices > ma)
    df.loc[buy_condition, 'signal'] = 1
    
    # SELL: RSI > threshold
    sell_condition = rsi > cfg.rsi_exit_threshold
    df.loc[sell_condition, 'signal'] = -1
    
    return df

def generate_signals_parallel(universe: List[str], price_data: Dict[str, pd.Series],
                              market_caps: Dict[str, float], cfg: BacktestConfig,
                              strategy_type: str = "ps_ratio") -> Dict[str, pd.DataFrame]:
    """
    Genera segnali in parallelo per tutto l'universo
    
    Args:
        strategy_type: "ps_ratio" o "rsi2"
    """
    print(f"Generazione segnali {strategy_type} per {len(universe)} titoli...")
    
    signal_map = {}
    
    for ticker in tqdm(universe, desc=f"Segnali {strategy_type}"):
        if strategy_type == "rsi2":
            signals = generate_rsi2_signal(ticker, price_data, market_caps, cfg)
        else:
            signals = generate_ps_ratio_signal(ticker, price_data, market_caps, cfg)
        
        if signals is not None and len(signals) > 0:
            signal_map[ticker] = signals
    
    print(f"  Generati segnali per {len(signal_map)} titoli")
    return signal_map

# ========================= UNIVERSE SELECTION ===============================

def select_all_available(price_data: Dict[str, pd.Series], start_date: pd.Timestamp,
                        end_date: pd.Timestamp, cfg: BacktestConfig) -> List[str]:
    """Seleziona tutti i ticker con dati sufficienti"""
    universe = []
    for ticker, prices in price_data.items():
        mask = (prices.index >= start_date) & (prices.index <= end_date)
        if mask.sum() >= cfg.min_trading_days:
            universe.append(ticker)
    return universe

def select_top_market_cap(price_data: Dict[str, pd.Series], start_date: pd.Timestamp,
                          end_date: pd.Timestamp, cfg: BacktestConfig, top_n: int = 50) -> List[str]:
    """Seleziona i top N ticker per market cap"""
    # Fetch market caps
    _, market_caps = fetch_yf_fundamentals_parallel(list(price_data.keys()))
    
    # Filter valid tickers
    valid_tickers = []
    for ticker in price_data.keys():
        if ticker in market_caps:
            mcap_billions = market_caps[ticker] / 1e9
            if mcap_billions >= cfg.min_market_cap_billion:
                mask = (price_data[ticker].index >= start_date) & (price_data[ticker].index <= end_date)
                if mask.sum() >= cfg.min_trading_days:
                    valid_tickers.append((ticker, market_caps[ticker]))
    
    # Sort by market cap
    valid_tickers.sort(key=lambda x: x[1], reverse=True)
    
    return [t[0] for t in valid_tickers[:top_n]]

def select_rsi2_candidates(price_data: Dict[str, pd.Series], start_date: pd.Timestamp,
                           end_date: pd.Timestamp, cfg: BacktestConfig) -> List[str]:
    """
    Seleziona candidati per RSI(2) strategy
    Preferibilmente large cap liquidi
    """
    _, market_caps = fetch_yf_fundamentals_parallel(list(price_data.keys()))
    
    candidates = []
    for ticker in price_data.keys():
        if ticker in market_caps:
            mcap_billions = market_caps[ticker] / 1e9
            if mcap_billions >= cfg.min_market_cap_billion:
                mask = (price_data[ticker].index >= start_date) & (price_data[ticker].index <= end_date)
                if mask.sum() >= cfg.min_trading_days + cfg.rsi_ma_period:
                    candidates.append((ticker, market_caps[ticker]))
    
    # Sort by market cap and take top candidates
    candidates.sort(key=lambda x: x[1], reverse=True)
    
    return [t[0] for t in candidates[:100]]  # Top 100 per RSI

# ========================= CONTINUOUS BACKTEST ===============================

def continuous_backtest(signal_map: Dict[str, pd.DataFrame], cfg: BacktestConfig,
                       strategy_name: str = "STRATEGY", is_rsi2_strategy: bool = False) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Backtest continuo con gestione commissioni corrette e export CSV
    
    Returns:
        equity_df: DataFrame con equity curve
        trades_df: DataFrame con tutti i trades (per export CSV)
    """
    print(f"\n=== BACKTEST CONTINUO: {strategy_name} ===")
    
    # Trova date comuni
    all_dates = set()
    for df in signal_map.values():
        all_dates.update(df.index)
    
    all_dates = sorted(all_dates)
    if len(all_dates) == 0:
        return pd.DataFrame(), pd.DataFrame()
    
    # Initialize tracking
    cash = cfg.initial_capital
    positions = {}  # {ticker: {'shares': float, 'entry_price': float, 'entry_date': date, 'entry_value': float}}
    
    equity_curve = []
    trades_list = []  # Lista di tutti i trades per CSV
    
    print(f"Simulazione dal {all_dates[0].date()} al {all_dates[-1].date()}")
    print(f"Capitale iniziale: ${cash:,.2f}")
    print(f"Commissioni: ${cfg.commission_per_trade_usd} fisso + {cfg.commission_per_trade_pct*100}%")
    
    for current_date in tqdm(all_dates, desc=f"Backtest {strategy_name}"):
        # Get signals for today
        buy_signals = []
        sell_signals = []
        
        for ticker, sig_df in signal_map.items():
            if current_date not in sig_df.index:
                continue
            
            signal = sig_df.loc[current_date, 'signal']
            price = sig_df.loc[current_date, 'price']
            
            if pd.isna(price) or price <= 0:
                continue
            
            if signal == 1:  # BUY
                buy_signals.append((ticker, price))
            elif signal == -1:  # SELL
                sell_signals.append((ticker, price))
        
        # STEP 1: Process SELL signals first
        for ticker, sell_price in sell_signals:
            if ticker not in positions:
                continue
            
            pos = positions[ticker]
            shares = pos['shares']
            entry_price = pos['entry_price']
            entry_date = pos['entry_date']
            entry_value = pos['entry_value']
            
            # Calcola valore di vendita
            sell_value = shares * sell_price
            
            # Calcola commissione di vendita
            sell_commission = calculate_commission(sell_value, cfg)
            
            # Cash ricevuto (valore vendita - commissione)
            cash_received = sell_value - sell_commission
            
            # P&L del trade (cash ricevuto - capitale investito iniziale)
            entry_commission = pos.get('entry_commission', 0.0)
            total_invested = entry_value + entry_commission  # Quanto abbiamo speso in totale all'entrata
            pnl = cash_received - total_invested
            pnl_pct = (pnl / total_invested) * 100 if total_invested > 0 else 0.0
            
            # Holding period
            holding_days = (current_date - entry_date).days
            
            # Update cash
            cash += cash_received
            
            # Record trade
            trades_list.append({
                'strategy': strategy_name,
                'ticker': ticker,
                'entry_date': entry_date,
                'entry_price': entry_price,
                'entry_shares': shares,
                'entry_value': entry_value,
                'entry_commission': entry_commission,
                'exit_date': current_date,
                'exit_price': sell_price,
                'exit_value': sell_value,
                'exit_commission': sell_commission,
                'holding_days': holding_days,
                'pnl_usd': pnl,
                'pnl_pct': pnl_pct,
                'total_commissions': entry_commission + sell_commission
            })
            
            # Remove position
            del positions[ticker]
        
        # RSI(2) specific: Check for max holding period and profit target
        if is_rsi2_strategy:
            tickers_to_close = []
            for ticker, pos in positions.items():
                holding_days = (current_date - pos['entry_date']).days
                
                # Get current price
                if ticker not in signal_map or current_date not in signal_map[ticker].index:
                    continue
                
                current_price = signal_map[ticker].loc[current_date, 'price']
                if pd.isna(current_price) or current_price <= 0:
                    continue
                
                # Check profit target
                current_value = pos['shares'] * current_price
                entry_total = pos['entry_value'] + pos.get('entry_commission', 0.0)
                current_pnl_pct = ((current_value - entry_total) / entry_total) * 100
                
                # Close if max holding days or profit target hit
                if holding_days >= cfg.rsi_max_holding_days or current_pnl_pct >= cfg.rsi_profit_target_pct:
                    tickers_to_close.append((ticker, current_price))
            
            # Close these positions
            for ticker, exit_price in tickers_to_close:
                pos = positions[ticker]
                shares = pos['shares']
                entry_value = pos['entry_value']
                entry_commission = pos.get('entry_commission', 0.0)
                
                # Calculate exit
                exit_value = shares * exit_price
                exit_commission = calculate_commission(exit_value, cfg)
                cash_received = exit_value - exit_commission
                
                # P&L
                total_invested = entry_value + entry_commission
                pnl = cash_received - total_invested
                pnl_pct = (pnl / total_invested) * 100 if total_invested > 0 else 0.0
                
                holding_days = (current_date - pos['entry_date']).days
                
                cash += cash_received
                
                # Record trade
                trades_list.append({
                    'strategy': strategy_name,
                    'ticker': ticker,
                    'entry_date': pos['entry_date'],
                    'entry_price': pos['entry_price'],
                    'entry_shares': shares,
                    'entry_value': entry_value,
                    'entry_commission': entry_commission,
                    'exit_date': current_date,
                    'exit_price': exit_price,
                    'exit_value': exit_value,
                    'exit_commission': exit_commission,
                    'holding_days': holding_days,
                    'pnl_usd': pnl,
                    'pnl_pct': pnl_pct,
                    'total_commissions': entry_commission + exit_commission,
                    'exit_reason': 'max_holding' if holding_days >= cfg.rsi_max_holding_days else 'profit_target'
                })
                
                del positions[ticker]
        
        # STEP 2: Process BUY signals
        if cfg.accumulate_daily and len(buy_signals) > 0:
            # Calculate max allocation per ticker
            total_equity = cash + sum(p['shares'] * signal_map[t].loc[current_date, 'price'] 
                                     for t, p in positions.items() 
                                     if t in signal_map and current_date in signal_map[t].index)
            
            max_per_ticker = total_equity * cfg.max_alloc_per_ticker_frac
            
            for ticker, buy_price in buy_signals:
                # Skip if already have position
                if ticker in positions:
                    continue
                
                # Calculate shares to buy
                target_value = min(cfg.per_trade_usd, max_per_ticker)
                
                if target_value > cash:
                    continue  # Not enough cash
                
                # Calculate commission for this buy
                buy_commission = calculate_commission(target_value, cfg)
                
                # Total cash needed (value + commission)
                total_cash_needed = target_value + buy_commission
                
                if total_cash_needed > cash:
                    continue  # Not enough cash including commissions
                
                # Buy shares
                if cfg.fractional_shares:
                    shares = target_value / buy_price
                else:
                    shares = int(target_value / buy_price)
                
                if shares <= 0:
                    continue
                
                # Actual value bought
                actual_value = shares * buy_price
                
                # Deduct cash (value + commission)
                cash -= (actual_value + buy_commission)
                
                # Create position
                positions[ticker] = {
                    'shares': shares,
                    'entry_price': buy_price,
                    'entry_date': current_date,
                    'entry_value': actual_value,
                    'entry_commission': buy_commission
                }
        
        # Calculate portfolio value
        portfolio_value = cash
        for ticker, pos in positions.items():
            if ticker in signal_map and current_date in signal_map[ticker].index:
                current_price = signal_map[ticker].loc[current_date, 'price']
                if not pd.isna(current_price) and current_price > 0:
                    portfolio_value += pos['shares'] * current_price
        
        equity_curve.append({
            'date': current_date,
            'equity': portfolio_value,
            'cash': cash,
            'positions_count': len(positions)
        })
    
    # Create DataFrames
    equity_df = pd.DataFrame(equity_curve)
    if not equity_df.empty:
        equity_df.set_index('date', inplace=True)
    
    trades_df = pd.DataFrame(trades_list)
    
    # Export to CSV if enabled
    if cfg.export_trades_csv and not trades_df.empty:
        os.makedirs(cfg.csv_output_dir, exist_ok=True)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        csv_filename = f"{strategy_name}_{timestamp}_trades.csv"
        csv_path = os.path.join(cfg.csv_output_dir, csv_filename)
        trades_df.to_csv(csv_path, index=False)
        print(f"✓ Trades exportati in: {csv_path}")
    
    print(f"Backtest completato: {len(trades_list)} trades")
    
    return equity_df, trades_df

def buy_hold_benchmark_continuous(start_date: str, end_date: str, initial_capital: float) -> pd.DataFrame:
    """Buy & Hold SPY benchmark"""
    print(f"\n=== BUY & HOLD SPY ===")
    try:
        # Download SPY
        spy_data = yf.download("SPY", start=start_date, end=end_date, progress=False, auto_adjust=True)
        
        if spy_data.empty:
            return pd.DataFrame()
        
        close = spy_data["Close"]
        close.index = pd.to_datetime(close.index).tz_localize(None)
        
        # Buy at first price
        first_price = close.iloc[0]
        shares = initial_capital / first_price
        
        # Calculate equity curve
        equity = shares * close
        
        equity_df = pd.DataFrame({
            'equity': equity,
            'cash': 0.0,
            'positions_count': 1
        })
        
        return equity_df
        
    except Exception as e:
        print(f"Errore SPY benchmark: {e}")
        return pd.DataFrame()

# ========================= PLOTTING ===============================

def plot_portfolio_equity_and_drawdown(all_results: Dict):
    """Plot equity curves and drawdowns for all strategies"""
    fig = plt.figure(figsize=(16, 10))
    gs = GridSpec(3, 1, figure=fig, height_ratios=[2, 1, 1], hspace=0.3)
    
    ax_equity = fig.add_subplot(gs[0])
    ax_dd = fig.add_subplot(gs[1])
    ax_stats = fig.add_subplot(gs[2])
    ax_stats.axis('off')
    
    colors = {
        'approach_a': 'blue',
        'approach_b': 'green',
        'approach_c': 'orange',
        'approach_d': 'red'
    }
    
    labels = {
        'approach_a': 'A - All Stocks (P/S)',
        'approach_b': 'B - Top 50 (P/S)',
        'approach_c': 'C - Buy & Hold SPY',
        'approach_d': 'D - RSI(2) Strategy'
    }
    
    # Plot equity curves
    for approach_key, result in all_results.items():
        if result and 'equity' in result and not result['equity'].empty:
            equity = result['equity']['equity']
            ax_equity.plot(equity.index, equity.values, 
                          label=labels.get(approach_key, approach_key),
                          color=colors.get(approach_key, 'gray'),
                          linewidth=2)
            
            # Calculate and plot drawdown
            dd = calculate_drawdown(equity)
            ax_dd.plot(dd.index, dd.values,
                      label=labels.get(approach_key, approach_key),
                      color=colors.get(approach_key, 'gray'),
                      linewidth=1.5)
    
    ax_equity.set_title('Portfolio Equity Curves (2021-2025)', fontsize=14, fontweight='bold')
    ax_equity.set_ylabel('Portfolio Value ($)', fontsize=12)
    ax_equity.legend(loc='upper left', fontsize=10)
    ax_equity.grid(True, alpha=0.3)
    
    ax_dd.set_title('Drawdown (%)', fontsize=14, fontweight='bold')
    ax_dd.set_ylabel('Drawdown %', fontsize=12)
    ax_dd.set_xlabel('Date', fontsize=12)
    ax_dd.legend(loc='lower left', fontsize=10)
    ax_dd.grid(True, alpha=0.3)
    ax_dd.fill_between(ax_dd.get_xlim(), 0, ax_dd.get_ylim()[0], alpha=0.1, color='red')
    
    # Stats table
    stats_text = "PERFORMANCE SUMMARY\n" + "="*80 + "\n\n"
    
    for approach_key in ['approach_a', 'approach_b', 'approach_c', 'approach_d']:
        result = all_results.get(approach_key)
        if not result:
            continue
        
        stats_text += f"{labels.get(approach_key, approach_key)}:\n"
        stats_text += f"  Total Return: {result['total_return']*100:+.1f}%  |  "
        stats_text += f"Annualized: {result['annualized_return']*100:.1f}%  |  "
        stats_text += f"Sharpe: {result['sharpe_ratio']:.2f}  |  "
        stats_text += f"Max DD: {result['max_drawdown']:.1f}%  |  "
        stats_text += f"Trades: {result['trades_count']}\n\n"
    
    ax_stats.text(0.05, 0.95, stats_text, 
                 transform=ax_stats.transAxes,
                 fontsize=10,
                 verticalalignment='top',
                 fontfamily='monospace',
                 bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))
    
    plt.tight_layout()
    
    # Save plot
    os.makedirs('backtest_results', exist_ok=True)
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    plot_filename = f"backtest_results/comparison_{timestamp}.png"
    plt.savefig(plot_filename, dpi=150, bbox_inches='tight')
    print(f"\n✓ Grafico salvato: {plot_filename}")
    
    plt.show()

# ========================= MAIN COMPARISON ===============================

def run_comprehensive_comparison(cfg: BacktestConfig):
    """Run complete comparison of all strategies"""
    
    start_date = f"{cfg.start_year}-01-01"
    end_date = today_str()
    
    print(f"\n{'='*80}")
    print(f"COMPREHENSIVE BACKTEST COMPARISON")
    print(f"{'='*80}")
    print(f"Periodo: {start_date} -> {end_date}")
    print(f"Capitale iniziale: ${cfg.initial_capital:,.0f}")
    print(f"Commissioni: ${cfg.commission_per_trade_usd} + {cfg.commission_per_trade_pct*100}%")
    print(f"{'='*80}\n")
    
    # 1. Download data
    universe = get_realistic_universe()
    all_price_data = fetch_prices_parallel(universe, start_date, end_date)
    
    if len(all_price_data) < 10:
        print("Troppo pochi dati scaricati, uscita")
        return {}
    
    fundamentals_data, market_caps = fetch_yf_fundamentals_parallel(list(all_price_data.keys()))
    
    # 2. Define universes
    full_period_start = pd.to_datetime(start_date)
    full_period_end = pd.to_datetime(end_date)
    
    universe_all = select_all_available(all_price_data, full_period_start, full_period_end, cfg)
    universe_top50 = select_top_market_cap(all_price_data, full_period_start, full_period_end, cfg, top_n=50)
    universe_rsi2 = select_rsi2_candidates(all_price_data, full_period_start, full_period_end, cfg)
    
    print(f"\n=== UNIVERSI SELEZIONATI ===")
    print(f"Universe All: {len(universe_all)} titoli")
    print(f"Universe Top50: {len(universe_top50)} titoli")
    print(f"Universe RSI(2): {len(universe_rsi2)} titoli")
    
    # 3. Run backtests
    all_results = {}
    all_trades_dfs = {}  # Store trades DataFrames for analysis
    
    # Approach A & B: P/S strategies
    for approach_name, universe in [("approach_a", universe_all), ("approach_b", universe_top50)]:
        if len(universe) < 5:
            print(f"\n{approach_name.upper()}: troppi pochi titoli, skip")
            all_results[approach_name] = None
            continue
        
        signal_map = generate_signals_parallel(universe, all_price_data, market_caps, cfg, "ps_ratio")
        
        if len(signal_map) < 3:
            print(f"\n{approach_name.upper()}: troppi pochi segnali, skip")
            all_results[approach_name] = None
            continue
        
        equity, trades = continuous_backtest(signal_map, cfg, approach_name.upper(), is_rsi2_strategy=False)
        
        all_trades_dfs[approach_name] = trades  # Store for analysis
        
        eq = equity["equity"].dropna()
        if len(eq) > 1:
            total_return = eq.iloc[-1] / eq.iloc[0] - 1.0
            years = (eq.index[-1] - eq.index[0]).days / 365.25
            annualized_return = (eq.iloc[-1] / eq.iloc[0]) ** (1/years) - 1.0
            sharpe_ratio = calculate_sharpe_ratio(eq, cfg.risk_free_rate)
            max_drawdown = calculate_max_drawdown(eq)
        else:
            total_return = annualized_return = sharpe_ratio = max_drawdown = 0.0
        
        all_results[approach_name] = {
            'universe_size': len(universe),
            'signals_count': len(signal_map),
            'total_return': total_return,
            'annualized_return': annualized_return,
            'sharpe_ratio': sharpe_ratio,
            'max_drawdown': max_drawdown,
            'trades_count': len(trades),
            'equity': equity
        }
    
    # Approach D - RSI(2) Strategy
    print(f"\n=== RSI(2) STRATEGY (LARRY CONNORS) ===")
    if len(universe_rsi2) >= 5:
        signal_map_rsi2 = generate_signals_parallel(universe_rsi2, all_price_data, market_caps, cfg, "rsi2")
        
        if len(signal_map_rsi2) >= 3:
            equity_rsi2, trades_rsi2 = continuous_backtest(signal_map_rsi2, cfg, "APPROACH_D", is_rsi2_strategy=True)
            
            all_trades_dfs['approach_d'] = trades_rsi2  # Store for analysis
            
            eq_rsi2 = equity_rsi2["equity"].dropna()
            if len(eq_rsi2) > 1:
                total_return_rsi2 = eq_rsi2.iloc[-1] / eq_rsi2.iloc[0] - 1.0
                years_rsi2 = (eq_rsi2.index[-1] - eq_rsi2.index[0]).days / 365.25
                annualized_return_rsi2 = (eq_rsi2.iloc[-1] / eq_rsi2.iloc[0]) ** (1/years_rsi2) - 1.0
                sharpe_ratio_rsi2 = calculate_sharpe_ratio(eq_rsi2, cfg.risk_free_rate)
                max_drawdown_rsi2 = calculate_max_drawdown(eq_rsi2)
            else:
                total_return_rsi2 = annualized_return_rsi2 = sharpe_ratio_rsi2 = max_drawdown_rsi2 = 0.0
            
            all_results['approach_d'] = {
                'universe_size': len(universe_rsi2),
                'signals_count': len(signal_map_rsi2),
                'total_return': total_return_rsi2,
                'annualized_return': annualized_return_rsi2,
                'sharpe_ratio': sharpe_ratio_rsi2,
                'max_drawdown': max_drawdown_rsi2,
                'trades_count': len(trades_rsi2),
                'equity': equity_rsi2
            }
        else:
            print(f"APPROACH_D: troppi pochi segnali RSI(2), skip")
            all_results['approach_d'] = None
    else:
        print(f"APPROACH_D: troppi pochi titoli per RSI(2), skip")
        all_results['approach_d'] = None
    
    # Approach C: Buy & Hold SPY
    print(f"\n=== BUY & HOLD SPY ===")
    spy_equity = buy_hold_benchmark_continuous(start_date, end_date, cfg.initial_capital)
    if not spy_equity.empty:
        eq_spy = spy_equity["equity"].dropna()
        if len(eq_spy) > 1:
            spy_total_return = eq_spy.iloc[-1] / eq_spy.iloc[0] - 1.0
            years = (eq_spy.index[-1] - eq_spy.index[0]).days / 365.25
            spy_annualized = (eq_spy.iloc[-1] / eq_spy.iloc[0]) ** (1/years) - 1.0
            spy_sharpe = calculate_sharpe_ratio(eq_spy, cfg.risk_free_rate)
            spy_max_dd = calculate_max_drawdown(eq_spy)
        else:
            spy_total_return = spy_annualized = spy_sharpe = spy_max_dd = 0.0
        
        all_results['approach_c'] = {
            'universe_size': 1,
            'signals_count': 1,
            'total_return': spy_total_return,
            'annualized_return': spy_annualized,
            'sharpe_ratio': spy_sharpe,
            'max_drawdown': spy_max_dd,
            'trades_count': 1,
            'equity': spy_equity
        }
    else:
        all_results['approach_c'] = None
    
    # 4. ANALISI TRADES DA CSV
    print(f"\n{'='*80}")
    print("ANALISI DETTAGLIATA TRADES")
    print(f"{'='*80}")
    
    for approach_key, trades_df in all_trades_dfs.items():
        if trades_df is not None and not trades_df.empty:
            # Find the CSV file
            strategy_name = approach_key.upper()
            csv_files = [f for f in os.listdir(cfg.csv_output_dir) if f.startswith(strategy_name) and f.endswith('.csv')]
            
            if csv_files:
                csv_path = os.path.join(cfg.csv_output_dir, csv_files[-1])  # Most recent
                analysis = analyze_trades_from_csv(csv_path, cfg)
                print_trade_analysis(analysis, strategy_name)
    
    # 5. Plot results
    print(f"\n{'='*80}")
    print("GENERAZIONE GRAFICI")
    print(f"{'='*80}")
    
    plot_portfolio_equity_and_drawdown(all_results)
    
    # 6. Final summary
    print(f"\n{'='*80}")
    print("CONFRONTO FINALE STRATEGIE")
    print(f"{'='*80}")
    
    approaches = {
        'A - Tutti i titoli (P/S)': all_results.get('approach_a'),
        'B - Top 50 market cap (P/S)': all_results.get('approach_b'),
        'C - Buy & Hold SPY': all_results.get('approach_c'),
        'D - RSI(2) Strategy (Connors)': all_results.get('approach_d')
    }
    
    for name, result in approaches.items():
        if not result:
            print(f"\n{name}: Dati non disponibili")
            continue
        
        print(f"\n{name}:")
        print(f"  Total Return:              {result['total_return']*100:+.1f}%")
        print(f"  Annualized Return:         {result['annualized_return']*100:.1f}%")
        print(f"  Sharpe Ratio:              {result['sharpe_ratio']:.2f}")
        print(f"  Max Drawdown:              {result['max_drawdown']:.1f}%")
        print(f"  Total Trades:              {result['trades_count']}")
        print(f"  Universe Size:             {result['universe_size']}")
    
    return all_results

# ====================== MAIN ==========================================

if __name__ == "__main__":
    cfg = BacktestConfig(
        initial_capital=10000.0,
        per_trade_usd=800.0,
        start_year=2022,  # CORREZIONE: Inizia dal 2022 dove abbiamo dati fondamentali
        rolling_window=30,
        min_periods=30,
        entry_k_std=-0.95,
        exit_k_std=0.5,
        accumulate_daily=True,
        fractional_shares=True,
        max_alloc_per_ticker_frac=0.25,
        rebalance_frequency_years=0.5,
        min_trading_days=10,
        min_market_cap_billion=1.0,
        risk_free_rate=0.02,
        
        # Commissioni (modifica questi valori secondo il tuo broker)
        commission_per_trade_usd=1.0,      # Es: $1 per trade
        commission_per_trade_pct=0.001,    # Es: 0.1% per trade
        
        # Larry Connors RSI(2) parameters
        rsi_period=2,
        rsi_entry_threshold=20,
        rsi_exit_threshold=90,
        rsi_ma_period=200,
        rsi_ma_type="SMA",
        rsi_max_holding_days=10,
        rsi_profit_target_pct=10.0,
        
        # CSV export
        export_trades_csv=True,
        csv_output_dir="backtest_results"
    )
    
    print(f"\n{'='*80}")
    print("BACKTEST COMPLETO CON ANALISI TRADES")
    print(f"{'='*80}")
    print(f"\nConfigurazioni:")
    print(f"  - Capitale iniziale: ${cfg.initial_capital:,.0f}")
    print(f"  - Commissioni: ${cfg.commission_per_trade_usd} fisso + {cfg.commission_per_trade_pct*100}%")
    print(f"  - RSI Period: {cfg.rsi_period}")
    print(f"  - RSI Entry: < {cfg.rsi_entry_threshold}")
    print(f"  - RSI Exit: > {cfg.rsi_exit_threshold}")
    print(f"  - MA Filter: {cfg.rsi_ma_period}-day {cfg.rsi_ma_type}")
    print(f"  - Max holding: {cfg.rsi_max_holding_days} days")
    print(f"  - Profit target: {cfg.rsi_profit_target_pct}%")
    print(f"  - Workers: {MAX_WORKERS}")
    print(f"  - Export CSV: {'ON' if cfg.export_trades_csv else 'OFF'}")
    print(f"\nPERIODO: {cfg.start_year}-2025")
    print(f"{'='*80}\n")
    
    start_time = time.time()
    results = run_comprehensive_comparison(cfg)
    end_time = time.time()
    
    print(f"\n{'='*80}")
    print(f"BACKTEST COMPLETATO IN {end_time - start_time:.1f} SECONDI")
    print(f"{'='*80}")
    
    print("\n✓ Risultati salvati in: backtest_results/")
    print("  - Grafici equity e drawdown")
    print("  - CSV trades per ogni strategia")
    print("  - Analisi dettagliata trades")
```
