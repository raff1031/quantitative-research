```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import glob
import os
from scipy.cluster import hierarchy
from scipy.spatial.distance import squareform

# ========================================
# 0. CARICAMENTO CSV CON ESTRAZIONE TICKER
# ========================================

path = r'C:\Users\roman\Desktop\quant\csv strats\Nuova cartella\*.csv'
all_files = glob.glob(path)

print(f"📁 Trovati {len(all_files)} file CSV\n")

def estrai_ticker_da_filename(filepath):
    """
    Estrae il ticker dal nome del file dopo 'vs'
    Es: 'correlation 10Y trasury vs NQ nt8 range 0.5 vwap ema 12.csv' -> 'NQ'
    """
    filename = os.path.basename(filepath)
    
    if ' vs ' in filename.lower():
        parts = filename.lower().split(' vs ')
        if len(parts) > 1:
            after_vs = parts[1]
            ticker = after_vs.split()[0].upper()
            return ticker
    
    return filename.replace('.csv', '')[:10]

# Carica tutti i CSV e aggiungi la colonna Ticker
dfs = []
ticker_count = {}

for file in all_files:
    try:
        temp_df = pd.read_csv(file, sep=';')
        ticker = estrai_ticker_da_filename(file)
        temp_df['Ticker'] = ticker
        ticker_count[ticker] = ticker_count.get(ticker, 0) + 1
        dfs.append(temp_df)
        print(f"   ✓ {os.path.basename(file)[:60]}... -> {ticker}")
    except Exception as e:
        print(f"   ✗ Errore: {file}: {e}")

df = pd.concat(dfs, ignore_index=True)

print(f"\n📊 Dataset totale: {len(df):,} righe")
print(f"🎯 Ticker estratti: {list(ticker_count.keys())}")

# ========================================
# 1. PULIZIA DATI
# ========================================

def pulisci_colonna_numerica(series):
    if series.dtype == 'object':
        cleaned = series.astype(str).str.replace('€', '').str.replace('$', '').str.replace(' ', '')
        if cleaned.str.contains(',').any() and cleaned.str.contains(r'\.').any():
            cleaned = cleaned.str.replace('.', '').str.replace(',', '.')
        elif cleaned.str.contains(',').any():
            cleaned = cleaned.str.replace(',', '.')
        return pd.to_numeric(cleaned, errors='coerce')
    return series

print("\n🧹 Pulizia dati in corso...")
colonne_numeriche = ['Net profit', 'Cum. net profit', 'Gross profit', 'Gross loss', 'Commission']

for col in colonne_numeriche:
    if col in df.columns:
        df[col] = pulisci_colonna_numerica(df[col])

df['Period'] = pd.to_datetime(df['Period'], errors='coerce')
df = df.dropna(subset=['Net profit', 'Period', 'Ticker'])
df = df.sort_values(['Period', 'Ticker']).reset_index(drop=True)

print(f"✅ Dati puliti: {len(df):,} righe valide")
print(f"💾 Memory usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# ========================================
# 2. ANALISI OVERALL (TUTTI I TRADE INSIEME)
# ========================================

print("\n" + "="*70)
print("📈 PARTE 1: OVERALL PERFORMANCE (TUTTI I TICKER COMBINATI)")
print("="*70)

def calcola_metriche_ottimizzato(group_df):
    """Calcola metriche per un gruppo di trade"""
    net_profits = group_df['Net profit'].values
    equity = np.cumsum(net_profits)
    
    # Maximum Drawdown
    running_max = np.maximum.accumulate(equity)
    drawdown_abs = equity - running_max
    max_drawdown_abs = np.min(drawdown_abs)
    
    with np.errstate(divide='ignore', invalid='ignore'):
        drawdown_pct = np.where(running_max != 0, 
                                (drawdown_abs / running_max) * 100, 
                                0)
    max_drawdown_pct = np.min(drawdown_pct)
    
    # Sharpe Ratio
    if len(net_profits) > 1 and np.std(net_profits) > 0:
        sharpe = np.mean(net_profits) / np.std(net_profits) * np.sqrt(252)
    else:
        sharpe = 0
    
    # Altre metriche
    largest_win = np.max(net_profits)
    largest_loss = np.min(net_profits)
    win_rate = np.sum(net_profits > 0) / len(net_profits) * 100
    
    total_wins = np.sum(net_profits[net_profits > 0])
    total_losses = np.abs(np.sum(net_profits[net_profits < 0]))
    profit_factor = total_wins / total_losses if total_losses > 0 else np.inf
    
    return {
        'Total Trades': len(group_df),
        'Total Profit': equity[-1],
        'Max Drawdown €': max_drawdown_abs,
        'Max Drawdown %': max_drawdown_pct,
        'Sharpe Ratio': sharpe,
        'Largest Win': largest_win,
        'Largest Loss': largest_loss,
        'Win Rate %': win_rate,
        'Profit Factor': profit_factor,
        'Avg Trade': np.mean(net_profits),
        'Equity': equity,
        'Drawdown_pct_series': drawdown_pct
    }

print("\n⏳ Calcolo metriche overall...")
metriche_overall = calcola_metriche_ottimizzato(df)
equity_overall = metriche_overall.pop('Equity')
drawdown_pct_overall = metriche_overall.pop('Drawdown_pct_series')

print("\n📊 METRICHE OVERALL:")
for key, value in metriche_overall.items():
    if 'Ratio' in key or '%' in key:
        print(f"{key:.<30} {value:.2f}")
    elif key == 'Total Trades':
        print(f"{key:.<30} {value:,}")
    else:
        print(f"{key:.<30} €{value:,.2f}")

# ========================================
# 3. GRAFICI OVERALL
# ========================================

print("\n📊 Generazione grafici overall...")

MAX_POINTS = 100000
if len(df) > MAX_POINTS:
    step = len(df) // MAX_POINTS
    plot_indices = np.arange(0, len(df), step)
else:
    plot_indices = np.arange(len(df))

dates_plot = df['Period'].iloc[plot_indices]
equity_plot = equity_overall[plot_indices]
drawdown_pct_plot = drawdown_pct_overall[plot_indices]

# Sample per istogramma
if len(df) > 50000:
    sample_trades = np.random.choice(df['Net profit'].values, size=50000, replace=False)
else:
    sample_trades = df['Net profit'].values

# --- FIGURA 1: Overall Performance ---
fig1, axes = plt.subplots(2, 2, figsize=(18, 12))

# Equity Line
ax1 = axes[0, 0]
ax1.plot(dates_plot, equity_plot, linewidth=2, color='#2E86AB', label='Equity Line')
ax1.fill_between(dates_plot, 0, equity_plot, alpha=0.3, color='#2E86AB')
ax1.axhline(y=0, color='red', linestyle='--', alpha=0.5, linewidth=1)
idx_max_dd = np.argmin(drawdown_pct_plot)
ax1.scatter(dates_plot.iloc[idx_max_dd], equity_plot[idx_max_dd], 
           color='red', s=150, zorder=5, marker='v', label='Max DD Point')
ax1.set_title(f'Overall Equity Line - Profit: €{equity_overall[-1]:,.2f}', 
              fontsize=14, fontweight='bold')
ax1.set_xlabel('Data')
ax1.set_ylabel('Profitto Cumulativo (€)')
ax1.grid(True, alpha=0.3)
ax1.legend()
ax1.ticklabel_format(style='plain', axis='y')

# Drawdown
ax2 = axes[0, 1]
ax2.fill_between(dates_plot, 0, drawdown_pct_plot, color='#A4031F', alpha=0.6)
ax2.plot(dates_plot, drawdown_pct_plot, linewidth=1.5, color='#8B0000')
ax2.set_title(f'Overall Drawdown % - Max: {metriche_overall["Max Drawdown %"]:.2f}%', 
              fontsize=14, fontweight='bold')
ax2.set_xlabel('Data')
ax2.set_ylabel('Drawdown (%)')
ax2.grid(True, alpha=0.3)

# Distribuzione Trade
ax3 = axes[1, 0]
ax3.hist(sample_trades, bins=100, color='#2E86AB', alpha=0.7, edgecolor='black')
ax3.axvline(x=0, color='red', linestyle='--', linewidth=2, label='Break Even')
ax3.axvline(x=np.mean(sample_trades), color='green', linestyle='--', 
            linewidth=2, label=f'Media: €{np.mean(sample_trades):.2f}')
ax3.set_title('Distribuzione Net Profit per Trade', fontsize=12, fontweight='bold')
ax3.set_xlabel('Net Profit (€)')
ax3.set_ylabel('Frequenza')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Boxplot
ax4 = axes[1, 1]
ax4.boxplot(sample_trades, vert=True, patch_artist=True,
            boxprops=dict(facecolor='#2E86AB', alpha=0.7),
            medianprops=dict(color='red', linewidth=2))
ax4.axhline(y=0, color='red', linestyle='--', linewidth=1)
ax4.set_title('Boxplot - Distribuzione Trade', fontsize=12, fontweight='bold')
ax4.set_ylabel('Net Profit (€)')
ax4.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('overall_performance.png', dpi=300, bbox_inches='tight')
plt.show()

print("   ✓ Grafico overall salvato: overall_performance.png")

# ========================================
# 4. ANALISI PER TICKER E CORRELAZIONE
# ========================================

print("\n" + "="*70)
print("🎯 PARTE 2: ANALISI PER TICKER E CORRELAZIONE")
print("="*70)

tickers_unici = sorted(df['Ticker'].unique())
print(f"\n🎯 Ticker unici: {len(tickers_unici)}")
print(f"   {tickers_unici}")

# Pivot table per correlazione
pivot_returns = df.pivot_table(
    index='Period',
    columns='Ticker',
    values='Net profit',
    aggfunc='sum'
)

print(f"\n📐 Pivot table shape: {pivot_returns.shape}")
print(f"📅 Date range: {pivot_returns.index.min().date()} to {pivot_returns.index.max().date()}")

pivot_returns_filled = pivot_returns.fillna(0)

# Calcolo correlazione
correlation_matrix = pivot_returns_filled.corr(method='pearson')

print("\n📊 MATRICE DI CORRELAZIONE:")
print(correlation_matrix.round(2))

# Statistiche correlazione
correlazioni_flat = correlation_matrix.values[np.triu_indices_from(correlation_matrix.values, k=1)]

if len(correlazioni_flat) > 0:
    print(f"\n📈 Statistiche Correlazione:")
    print(f"   Media:   {np.mean(correlazioni_flat):>6.3f}")
    print(f"   Mediana: {np.median(correlazioni_flat):>6.3f}")
    print(f"   Min:     {np.min(correlazioni_flat):>6.3f}")
    print(f"   Max:     {np.max(correlazioni_flat):>6.3f}")
    print(f"   Std:     {np.std(correlazioni_flat):>6.3f}")

# Top coppie
if len(tickers_unici) > 1:
    corr_pairs = []
    for i in range(len(correlation_matrix.columns)):
        for j in range(i+1, len(correlation_matrix.columns)):
            corr_pairs.append({
                'Ticker_1': correlation_matrix.columns[i],
                'Ticker_2': correlation_matrix.columns[j],
                'Correlation': correlation_matrix.iloc[i, j]
            })
    
    corr_pairs_df = pd.DataFrame(corr_pairs).sort_values('Correlation', ascending=False)
    
    print("\n🔥 TOP 5 COPPIE PIÙ CORRELATE:")
    print(corr_pairs_df.head().to_string(index=False))
    
    print("\n❄️ TOP 5 COPPIE MENO CORRELATE:")
    print(corr_pairs_df.tail().to_string(index=False))

# ========================================
# 5. GRAFICI CORRELAZIONE
# ========================================

print("\n📊 Generazione grafici correlazione...")

fig2 = plt.figure(figsize=(20, 14))
gs = fig2.add_gridspec(3, 2, hspace=0.3, wspace=0.3, height_ratios=[2, 1, 1])

# --- Heatmap Correlazione ---
ax1 = fig2.add_subplot(gs[0, :])

if len(tickers_unici) > 1:
    try:
        linkage = hierarchy.linkage(squareform(1 - correlation_matrix.abs()), method='average')
        dendro_order = hierarchy.dendrogram(linkage, no_plot=True)['leaves']
        correlation_matrix_clustered = correlation_matrix.iloc[dendro_order, dendro_order]
    except:
        correlation_matrix_clustered = correlation_matrix
    
    sns.heatmap(correlation_matrix_clustered, 
                annot=True, 
                fmt='.2f',
                cmap='RdYlGn',
                center=0,
                vmin=-1, 
                vmax=1,
                square=True,
                linewidths=0.5,
                cbar_kws={'label': 'Correlazione'},
                ax=ax1,
                annot_kws={'size': 9})
    ax1.set_title('Matrice di Correlazione tra Ticker (Hierarchical Clustering)', 
                  fontsize=16, fontweight='bold', pad=20)
else:
    ax1.text(0.5, 0.5, 'Serve almeno 2 ticker per la correlazione', 
             ha='center', va='center', fontsize=14)
    ax1.axis('off')

# --- Distribuzione Correlazioni ---
ax2 = fig2.add_subplot(gs[1, 0])

if len(correlazioni_flat) > 0:
    ax2.hist(correlazioni_flat, bins=min(30, len(correlazioni_flat)), 
             color='#2E86AB', alpha=0.7, edgecolor='black')
    ax2.axvline(x=0, color='red', linestyle='--', linewidth=2, label='Zero')
    ax2.axvline(x=np.mean(correlazioni_flat), color='green', linestyle='--', 
                linewidth=2, label=f'Media: {np.mean(correlazioni_flat):.2f}')
    ax2.set_title('Distribuzione delle Correlazioni', fontsize=13, fontweight='bold')
    ax2.set_xlabel('Correlazione')
    ax2.set_ylabel('Frequenza')
    ax2.legend()
    ax2.grid(True, alpha=0.3)

# --- Connessioni Forti ---
ax3 = fig2.add_subplot(gs[1, 1])

threshold = 0.3
strong_corr = correlation_matrix.copy()
strong_corr[abs(strong_corr) < threshold] = 0
connections = (abs(strong_corr) > 0).sum() - 1

colors = ['#A4031F' if c > np.median(connections) else '#2E86AB' for c in connections]
ax3.bar(range(len(connections)), connections.values, color=colors, alpha=0.7)
ax3.set_xticks(range(len(connections)))
ax3.set_xticklabels(connections.index, rotation=45, ha='right')
ax3.set_title(f'Correlazioni Forti (|r| > {threshold}) per Ticker', 
              fontsize=13, fontweight='bold')
ax3.set_xlabel('Ticker')
ax3.set_ylabel('Numero Connessioni')
ax3.grid(True, alpha=0.3, axis='y')

# --- Equity Lines per Ticker ---
ax4 = fig2.add_subplot(gs[2, 0])

for ticker in tickers_unici[:10]:
    equity = pivot_returns_filled[ticker].cumsum()
    ax4.plot(equity.index, equity.values, label=ticker, linewidth=2, alpha=0.8)

ax4.set_title('Equity Lines per Ticker (Top 10)', fontsize=13, fontweight='bold')
ax4.set_xlabel('Data')
ax4.set_ylabel('Profit Cumulativo (€)')
ax4.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=9)
ax4.grid(True, alpha=0.3)
ax4.axhline(y=0, color='red', linestyle='--', alpha=0.5)

# --- Volatilità vs Sharpe ---
ax5 = fig2.add_subplot(gs[2, 1])

vol_per_ticker = pivot_returns_filled.std()
mean_per_ticker = pivot_returns_filled.mean()
sharpe_per_ticker = (mean_per_ticker / vol_per_ticker * np.sqrt(252)).fillna(0)

scatter = ax5.scatter(vol_per_ticker, sharpe_per_ticker, 
                      s=200, alpha=0.6, c=range(len(vol_per_ticker)), 
                      cmap='viridis', edgecolors='black', linewidth=1.5)

for i, ticker in enumerate(vol_per_ticker.index):
    ax5.annotate(ticker, (vol_per_ticker[ticker], sharpe_per_ticker[ticker]),
                xytext=(5, 5), textcoords='offset points', fontsize=9, fontweight='bold')

ax5.axhline(y=0, color='red', linestyle='--', alpha=0.5)
ax5.set_title('Volatilità vs Sharpe Ratio per Ticker', fontsize=13, fontweight='bold')
ax5.set_xlabel('Volatilità (Std Net Profit)')
ax5.set_ylabel('Sharpe Ratio (annualizzato)')
ax5.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('correlation_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print("   ✓ Grafico correlazione salvato: correlation_analysis.png")

# ========================================
# 6. METRICHE DI DIVERSIFICAZIONE
# ========================================

print("\n" + "="*70)
print("🎯 METRICHE DI DIVERSIFICAZIONE")
print("="*70)

if len(tickers_unici) > 1:
    vol_individuali = pivot_returns_filled.std()
    vol_portafoglio = (pivot_returns_filled.mean(axis=1)).std()
    
    if vol_portafoglio > 0:
        diversification_ratio = vol_individuali.mean() / vol_portafoglio
        print(f"Diversification Ratio: {diversification_ratio:.2f}")
        print(f"  {'✅ Ottimo (>2)' if diversification_ratio > 2 else '⚠️ Buono (1-2)' if diversification_ratio > 1 else '❌ Scarso (<1)'}")
    
    avg_pairwise_corr = np.mean(correlazioni_flat)
    print(f"\nCorrelazione Media tra Coppie: {avg_pairwise_corr:.3f}")
    print(f"  {'✅ Bassa (<0.3)' if abs(avg_pairwise_corr) < 0.3 else '⚠️ Moderata (0.3-0.7)' if abs(avg_pairwise_corr) < 0.7 else '❌ Alta (>0.7)'}")
    
    negative_corr_count = (correlazioni_flat < 0).sum()
    total_pairs = len(correlazioni_flat)
    print(f"\nCoppie con Correlazione Negativa: {negative_corr_count}/{total_pairs} ({negative_corr_count/total_pairs*100:.1f}%)")
    print(f"  (Più alto = migliore hedging naturale)")

# ========================================
# 7. PERFORMANCE PER TICKER
# ========================================

print("\n" + "="*70)
print("📊 PERFORMANCE DETTAGLIATA PER TICKER")
print("="*70)

perf_per_ticker = []
for ticker in tickers_unici:
    df_ticker = df[df['Ticker'] == ticker]
    returns = df_ticker['Net profit'].values
    
    if len(returns) > 0:
        equity = np.cumsum(returns)
        running_max = np.maximum.accumulate(equity)
        drawdown = equity - running_max
        max_dd = np.min(drawdown)
        max_dd_pct = (max_dd / np.max(running_max) * 100) if np.max(running_max) > 0 else 0
        
        total_wins = np.sum(returns[returns > 0])
        total_losses = np.abs(np.sum(returns[returns < 0]))
        
        perf_per_ticker.append({
            'Ticker': ticker,
            'Trades': len(returns),
            'Total Profit': returns.sum(),
            'Avg Trade': returns.mean(),
            'Volatility': returns.std(),
            'Sharpe': (returns.mean() / returns.std() * np.sqrt(252)) if returns.std() > 0 else 0,
            'Win Rate %': (returns > 0).sum() / len(returns) * 100,
            'Profit Factor': total_wins / total_losses if total_losses > 0 else np.inf,
            'Max DD %': max_dd_pct,
            'Avg Corr': correlation_matrix[ticker].drop(ticker).mean() if len(tickers_unici) > 1 else 0
        })

perf_df = pd.DataFrame(perf_per_ticker).sort_values('Total Profit', ascending=False)
print(perf_df.to_string(index=False))

# ========================================
# 8. EXPORT
# ========================================

print("\n" + "="*70)
print("💾 ESPORTAZIONE RISULTATI")
print("="*70)

# Salva metriche overall
overall_export = pd.DataFrame([metriche_overall])
overall_export.to_csv('overall_metrics.csv', index=False)

# Salva correlazione
correlation_matrix.to_csv('correlation_matrix.csv')

# Salva performance per ticker
perf_df.to_csv('performance_per_ticker.csv', index=False)

# Salva coppie di correlazione
if len(tickers_unici) > 1:
    corr_pairs_df.to_csv('correlation_pairs.csv', index=False)

print("✅ File salvati:")
print("   - overall_metrics.csv")
print("   - overall_performance.png")
print("   - correlation_matrix.csv")
print("   - performance_per_ticker.csv")
print("   - correlation_analysis.png")
if len(tickers_unici) > 1:
    print("   - correlation_pairs.csv")

print("\n" + "="*70)
print("🎉 ANALISI COMPLETA TERMINATA!")
print("="*70)
```
